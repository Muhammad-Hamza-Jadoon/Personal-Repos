### adding 1 to any binary no

```python
from time import sleep
from IPython.display import clear_output

class TuringMachine:
    def __init__(self, tape_string):
        self.tape = list("_" * 10) + list(tape_string) + list("_" * 10)  # fixed!
        self.head = 10 + len(tape_string) - 1
        self.state = "q0"
        self.sleep_time = 1  # seconds per step

    def step(self):
        symbol = self.tape[self.head]
        if self.state == "q0":
            if symbol == "1":
                self.tape[self.head] = "0"
                self.head -= 1
                self.state = "q0"
            elif symbol == "0":
                self.tape[self.head] = "1"
                self.state = "HALT"
            elif symbol == "_":
                self.tape[self.head] = "1"
                self.state = "HALT"

    def visualize(self):
        clear_output(wait=True)
        tape_str = "".join(self.tape)
        pointer_str = " " * self.head + "↑"
        print(f"State: {self.state}")
        print(tape_str)
        print(pointer_str)

    def run(self):
        while self.state != "HALT":
            self.visualize()
            self.step()
            sleep(self.sleep_time)
        self.visualize()
        print("\n✅ Machine halted. Final binary:", self.get_output())

    def get_output(self):
        return ''.join(self.tape).strip("_")

# 👇 Run it
tm = TuringMachine("101111")
tm.run()
```
